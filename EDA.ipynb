```python
# ============================================================
# INSIDER THREAT DETECTION - EDA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

# If running in Colab, upload the CSV
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("=" * 70)
print("INSIDER THREAT DETECTION - EXPLORATORY DATA ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 2. BASIC DATASET INFORMATION
# ------------------------------------------------------------

print("\n1. DATASET SHAPE")
print("-" * 50)

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])


print("\n2. FIRST 5 RECORDS")
print("-" * 50)

display(df.head())


print("\n3. COLUMN NAMES")
print("-" * 50)

for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")


# ------------------------------------------------------------
# 3. DATA TYPES
# ------------------------------------------------------------

print("\n4. DATA TYPES")
print("-" * 50)

print(df.dtypes)

print("\nData type summary:")
print(df.dtypes.value_counts())


# ------------------------------------------------------------
# 4. MISSING VALUES
# ------------------------------------------------------------

print("\n5. MISSING VALUE ANALYSIS")
print("-" * 50)

missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values,
    "Missing Percentage": (
        missing.values / len(df) * 100
    )
})

missing_table = missing_table[
    missing_table["Missing Values"] > 0
]

if missing_table.empty:
    print("No missing values found.")

else:
    display(
        missing_table.sort_values(
            "Missing Values",
            ascending=False
        )
    )


# ------------------------------------------------------------
# 5. DUPLICATE RECORDS
# ------------------------------------------------------------

print("\n6. DUPLICATE RECORD ANALYSIS")
print("-" * 50)

duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)


# ------------------------------------------------------------
# 6. NUMERICAL FEATURES
# ------------------------------------------------------------

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("\n7. NUMERICAL FEATURES")
print("-" * 50)

print("Number of numerical columns:",
      len(numeric_columns))

print(numeric_columns)


# ------------------------------------------------------------
# 7. CATEGORICAL FEATURES
# ------------------------------------------------------------

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("\n8. CATEGORICAL FEATURES")
print("-" * 50)

print("Number of categorical columns:",
      len(categorical_columns))

print(categorical_columns)


# ------------------------------------------------------------
# 8. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\n9. STATISTICAL SUMMARY")
print("-" * 50)

display(
    df.describe()
)


# ------------------------------------------------------------
# 9. RISK LABEL DISTRIBUTION
# ------------------------------------------------------------

print("\n10. RISK LABEL DISTRIBUTION")
print("-" * 50)

print(
    df["RiskLabel"].value_counts()
)

print("\nPercentage distribution:")

print(
    df["RiskLabel"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ------------------------------------------------------------
# 10. RISK LABEL BAR CHART
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="RiskLabel"
)

plt.title(
    "Distribution of Risk Labels"
)

plt.xlabel(
    "Risk Label"
)

plt.ylabel(
    "Number of Records"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 11. RISK SCORE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="RiskScore",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Risk Score"
)

plt.xlabel(
    "Risk Score"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 12. RISK SCORE BY RISK LABEL
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="RiskLabel",
    y="RiskScore"
)

plt.title(
    "Risk Score by Risk Label"
)

plt.xlabel(
    "Risk Label"
)

plt.ylabel(
    "Risk Score"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 13. ACTIVITY FLAG ANALYSIS
# ------------------------------------------------------------

flag_columns = [
    col for col in df.columns
    if col.endswith("_flag")
]

print("\n11. ACTIVITY FLAG FEATURES")
print("-" * 50)

print("Number of activity flag columns:",
      len(flag_columns))

print(flag_columns)


# ------------------------------------------------------------
# 14. ACTIVITY FREQUENCY
# ------------------------------------------------------------

flag_counts = df[flag_columns].sum().sort_values(
    ascending=False
)

print("\nActivity flag frequencies:")
display(
    flag_counts.to_frame(
        name="Number of Flagged Records"
    )
)


# ------------------------------------------------------------
# 15. TOP ACTIVITY FLAGS
# ------------------------------------------------------------

top_flags = flag_counts.head(15)

plt.figure(figsize=(10, 7))

top_flags.sort_values().plot(
    kind="barh"
)

plt.title(
    "Top 15 Activity Flags"
)

plt.xlabel(
    "Number of Flagged Records"
)

plt.ylabel(
    "Activity"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 16. CORRELATION ANALYSIS
# ------------------------------------------------------------

print("\n12. CORRELATION ANALYSIS")
print("-" * 50)

correlation_columns = flag_columns + [
    "RiskScore"
]

correlation_matrix = df[
    correlation_columns
].corr()


plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Matrix of Activity Features and Risk Score"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 17. ACTIVITY FLAGS VS RISK LABEL
# ------------------------------------------------------------

print("\n13. ACTIVITY PATTERNS BY RISK LABEL")
print("-" * 50)

activity_by_risk = (
    df.groupby("RiskLabel")[flag_columns]
    .mean()
    .T
)

display(
    activity_by_risk
)


# ------------------------------------------------------------
# 18. HEATMAP OF ACTIVITY VS RISK
# ------------------------------------------------------------

plt.figure(figsize=(12, 10))

sns.heatmap(
    activity_by_risk,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title(
    "Average Activity Flag Rate by Risk Label"
)

plt.xlabel(
    "Risk Label"
)

plt.ylabel(
    "Activity Feature"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 19. HIGH-RISK RECORDS
# ------------------------------------------------------------

print("\n14. HIGH-RISK RECORD ANALYSIS")
print("-" * 50)

high_risk = df[
    df["RiskLabel"].astype(str).str.lower() == "high"
]

print(
    "Number of High-risk records:",
    len(high_risk)
)

print(
    "Percentage of High-risk records:",
    round(
        len(high_risk) / len(df) * 100,
        2
    ),
    "%"
)


# ------------------------------------------------------------
# 20. MEDIUM-RISK RECORDS
# ------------------------------------------------------------

medium_risk = df[
    df["RiskLabel"].astype(str).str.lower() == "medium"
]

print(
    "Number of Medium-risk records:",
    len(medium_risk)
)


# ------------------------------------------------------------
# 21. LOW-RISK RECORDS
# ------------------------------------------------------------

low_risk = df[
    df["RiskLabel"].astype(str).str.lower() == "low"
]

print(
    "Number of Low-risk records:",
    len(low_risk)
)


# ------------------------------------------------------------
# 22. FINAL EDA SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("EDA SUMMARY")
print("=" * 70)

print(
    "\nDataset size:",
    df.shape
)

print(
    "Numerical columns:",
    len(numeric_columns)
)

print(
    "Categorical columns:",
    len(categorical_columns)
)

print(
    "Missing values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Activity flag features:",
    len(flag_columns)
)

print(
    "Risk categories:",
    df["RiskLabel"].unique()
)

print("\nEDA COMPLETED SUCCESSFULLY.")
```
